```python
import gwinc
import matplotlib.pyplot as plt
import numpy as np
plt.style.use('tableau-colorblind10')
```

```python
freq = np.logspace(0,4,10000)

aligo_budget = gwinc.load_budget("../ifo/canonical/aLIGO.yaml", freq=freq)
aligo_trace = aligo_budget.run()

CE_budget = gwinc.load_budget("../ifo/canonical/CE1.yaml", freq=freq)
CE_trace = CE_budget.run()

OzCE_budget = gwinc.load_budget("../ifo/variant/OzCE-20.yaml", freq=freq)
OzCE_trace = OzCE_budget.run()
```

```python
plt.figure(figsize=(8,5), dpi=180)
plt.loglog(aligo_trace.freq, aligo_trace.asd, lw=5, label='aLIGO')
plt.loglog(CE_trace.freq, CE_trace.asd, lw=5, label='CE1')
plt.loglog(OzCE_trace.freq, OzCE_trace.asd, lw=5, label='OzCE-20km')
plt.xlabel('Frequency [Hz]')
plt.ylabel(r'Strain [1/$\sqrt{\mathrm{Hz}}$]')
plt.xlim(5,5e3)
plt.ylim(3e-25, 2e-22)
plt.grid(which='both', lw=0.5, ls='--')
plt.legend()
plt.show()
```
