```python
import gwinc
import numpy as np
```

```python
budget = gwinc.load_budget("aLIGO")

coatings_file = "../ifo/variant/coatings/default.yaml"
coatings_struct = gwinc.struct.Struct.from_file(coatings_file)

ifo_file = "../ifo/variant/OzCE-20.yaml"
ifo_struct = gwinc.struct.Struct.from_file(ifo_file)

ifo_struct.update(coatings_struct)

trace = budget.run(ifo=ifo_struct)
```

```python
trace.plot()
```
